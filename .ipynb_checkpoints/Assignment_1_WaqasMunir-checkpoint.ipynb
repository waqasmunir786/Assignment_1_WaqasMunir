## 7. Write a program which takes 5 inputs from user for different subject’s marks, total it and generate mark sheet using grades ?

```python

subject1 = float(input("Enter marks of English: "))
subject2 = float(input("Enter marks of Urdu: "))
subject3 = float(input("Enter marks of Math: "))
subject4 = float(input("Enter marks of Pak Studies: "))
subject5 = float(input("Enter marks of Islamic Study: "))

total_marks = subject1 + subject2 + subject3 + subject4 + subject5


percentage = (total_marks / 500) * 100

if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
elif percentage >= 50:
    grade = "E"
else:
    grade = "F"

print("\n--- Your Marks Sheet ---")
print("English:", subject1)
print("Urdu:", subject2)
print("Math:", subject3)
print("Pak Studies:", subject4)
print("Islamic Studies:", subject5)
print("Total Marks:", total_marks)
print("Percentage:", round(percentage, 2), "%")
print("Grade:", grade)
```

## 8. Write a program which take input from user and identify that the given number is even or odd?

```python
num1 = int(input("Enter Your  Numer "))
if num1 % 2==0:
    print("Even Number:", num1)
else:
    print("Odd Number:", num1)
```
